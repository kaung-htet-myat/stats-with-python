# SVM

Objective of SVM is to find the boundary hyperplane with the largest margin between closest points from two (or more) class regions. In order to define the boundry and find the support vectors at training time, it solves **dual optimization problem** to find the support vectors of the boundary.

## Dual Optimization problem
$Maximize: L(\alpha) = \sum_i(\alpha_i) - \frac{1}{2}*(\sum_i\sum_j(\alpha_i * \alpha_j * y_i * y_j * K(x_i, x_j)))$
- maximizes the alpha
- by distributing the alpha score using $\alpha_i * \alpha_j$ inside the penalty term
- by giving higher alpha to similar points between two classes near the boundary ($y_i * y_j$ will result in penalty -1 if different class)
- by giving lower alpha to similar points within each class ($y_i * y_j$ will result in penalty 1 if same class)
- *to cover higher dimensional data, similarity score is calculated using kernel functions K*
- points with alpha>0 will become support vectors

## At inference time
$score(u) = \sum_i(\alpha_i * y_i * K(x_i, u))+b$
- calculate similarity score between each support vector and inference x
    to check which support vectors are nearest to x
- multiply each similarity score with y of each support vector to assign label
- multiply each similarity score with alpha to assign influence of each support vector
